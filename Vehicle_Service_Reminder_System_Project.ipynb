### Vehicle Service Reminder System

```python
# Vehicle Service Checker

print("--- Vehicle Service Checker ---")

# vehicle type (fixed input)
while True:
    vehicle_type = input("Enter vehicle type (2W/4W): ").upper()
    if vehicle_type in ["2W", "4W"]:
        break
    else:
        print("Please enter only 2W or 4W")

# vehicle name based on type
if vehicle_type == "2W":
    vehicle_name = input("Enter vehicle (bike/scooty): ").lower()
    service_interval = 2000
elif vehicle_type == "4W":
    vehicle_name = input("Enter vehicle (car): ").lower()
    service_interval = 5000

model = input("Enter model name: ")

# km details
last_km = int(input("Enter last service km: "))
current_km = int(input("Enter current km: "))

# use interval
next_service = last_km + service_interval
print("Next service at:", next_service, "km")

if current_km >= next_service:
    print("Service is due. Please service your vehicle.")
else:
    remaining = next_service - current_km
    print("Service not due yet.")
    print("Remaining km:", remaining)

# save service details
save = input("Do you want to save service details? (yes/no): ").lower()

if save == "yes":
    date = input("Enter service date: ")
    work = input("Work done: ")
    extra = input("Any extra work: ")

    f = open("service.txt", "a")
    f.write("Vehicle Type: " + vehicle_type + "\n")
    f.write("Vehicle: " + vehicle_name + "\n")
    f.write("Model: " + model + "\n")
    f.write("Service Interval: " + str(service_interval) + " km\n")
    f.write("Date: " + date + "\n")
    f.write("Work: " + work + "\n")
    f.write("Extra: " + extra + "\n")
    f.write("----------------------\n")
    f.close()

    print("Details saved")

# view history
view = input("Do you want to see service history? (yes/no): ").lower()

if view == "yes":
    try:
        f = open("service.txt", "r")
        print("--- Service History ---")
        print(f.read())
        f.close()
    except:
        print("No records found")
```
